```python


import pandas as pd
import os


# ---------------------------------------------------
# INPUT FILES
# ---------------------------------------------------

final_file = "90/final_merged_output.csv"

loc_file = (
    "localization/"
    "all_genes_location_without_chx.csv"
)

output_file = (
    "90/final_merged_output_localization.csv"
)

# ---------------------------------------------------
# LOAD
# ---------------------------------------------------

final = pd.read_csv(
    final_file
)

loc = pd.read_csv(
    loc_file
)

print(
    "Final:",
    final.shape
)

print(
    "Localization:",
    loc.shape
)

# ---------------------------------------------------
# KEEP REQUIRED COLS
# ---------------------------------------------------

loc = loc[
    [
        "gene_old",
        "UniProt",
        "sequence",
        "localization"
    ]
].copy()

# sequence length
loc["seq_count"] = (
    loc["sequence"]
    .fillna("")
    .astype(str)
    .str.len()
)

# ---------------------------------------------------
# MERGE
# ---------------------------------------------------

merged = final.merge(
    loc,
    left_on="gene",
    right_on="gene_old",
    how="left"
)

# ---------------------------------------------------
# SELECT OUTPUT COLUMNS
# ---------------------------------------------------

avg_cols = [
    c
    for c in merged.columns
    if c.startswith("avg_")
]

keep = [
    "gene",
    "cds_length",
    "UniProt",
]

keep += avg_cols

keep += [
    "sequence",
    "localization",
    "seq_count"
]

keep = [
    c
    for c in keep
    if c in merged.columns
]

final_out = (
    merged[keep]
    .rename(
        columns={
            "seq_count": "Length"
        }
    )
)

# ---------------------------------------------------
# SAVE
# ---------------------------------------------------

final_out.to_csv(
    output_file,
    index=False
)

print(
    "\nSaved:",
    output_file
)

print(
    "Output shape:",
    final_out.shape
)

print(
    final_out.head()
)
```

```python
final_mito= final_out[final_out["localization"]=="mitochondrial"]
final_non= final_out[final_out["localization"]!="mitochondrial"]
```

```python
final_mito.to_csv("90/final_merged_output_mito.csv",index=False)
final_non.to_csv("90/final_merged_output_non.csv",index=False)
```
