```python

import h5py
import pandas as pd


# ============================================================
# INPUT FILES
# ============================================================

# ---------- INTERACTOME ----------
interactome_files = [
    "TOM_interactome_rep1.h5",
    "TOM_interactome_rep2.h5",
    "TOM_interactome_rep3.h5"
]

# ---------- TRANSLATOME ----------
translatome_files = [
    "Translatome_rep1.h5",
    "Translatome_rep2.h5",
    "Translatome_rep3.h5"
]

# ---------- OUTPUT ----------
OUTPUT_CSV = "all_genes_metadata_without_chx.csv"


# ============================================================
# ALL FILES
# ============================================================

all_files = interactome_files + translatome_files


# ============================================================
# EXTRACT ALL UNIQUE GENES
# ============================================================

gene_metadata = {}

for fname in all_files:

    print(f"\nProcessing: {fname}")

    with h5py.File(fname, "r") as f:

        for gene in f.keys():

            # Skip if already collected
            if gene in gene_metadata:
                continue

            ds = f[gene]

           
            protein_id = ds.attrs.get("protein_id", "")
           
            cds_length = ds.attrs.get("cds_length", 0)

            
            if isinstance(protein_id, bytes):
                protein_id = protein_id.decode()

          

            gene_metadata[gene] = {
                "gene": gene,
               
                "protein_id": protein_id,
              
                "cds_length": int(cds_length)
            }


# ============================================================
# CREATE DATAFRAME
# ============================================================

df = pd.DataFrame(
    list(gene_metadata.values())
)

# Sort by gene name
df = df.sort_values("gene")


# ============================================================
# SAVE CSV
# ============================================================

df.to_csv(
    OUTPUT_CSV,
    index=False
)

print(f"\nSaved CSV: {OUTPUT_CSV}")
print(f"Total genes: {len(df)}\n")
```
